# Holmes Inspect API Test
Use this notebook to validate your new `hm.inspect(df)` API on a realistic missingness dataset.

```python
import pandas as pd
import holmes as hm
```

```python
# Build a sample dataset with realistic missing-value behavior
n = 40
sample_df = pd.DataFrame({
    "customer_id": list(range(1001, 1001 + n)),
    "country": ["US", "DE", "FR", "BR", "IN"] * 8,
    "age": [
        24, 29, None, 31, 42,
        35, None, 28, 39, 26,
        44, 30, 27, None, 33,
        41, 25, 38, None, 36,
        29, 34, 32, 45, None,
        27, 37, 31, 40, 28,
        None, 26, 35, 39, 30,
        43, None, 33, 41, 29,
    ],
    "salary": [
        None, None, None, 62000, None,
        None, None, None, 58000, None,
        None, None, None, 61000, None,
        None, None, None, 60500, None,
        None, None, None, 59000, None,
        None, None, None, 64000, None,
        None, None, None, 60000, None,
        None, None, None, 61500, None,
    ],
    "segment": ["A", "B", "A", "C"] * 10,
    "last_purchase_days": [
        3, 7, 14, None, 10,
        5, 21, 8, 30, None,
        12, 17, 6, 9, 13,
        None, 20, 11, 15, 18,
        4, None, 22, 16, 19,
        2, 25, 27, None, 24,
        26, 23, 29, 28, None,
        1, 32, 31, 33, 34,
    ],
})

sample_df.head()
```

```python
# Main inspect test (rich output in Jupyter)
report = hm.inspect(sample_df)
report
```

```python
# Quick verification and plain-text output
report.show()

print("\nCritical columns:", [c["column_name"] for c in report.critical_columns])
print("Moderate columns:", [c["column_name"] for c in report.moderate_columns])
print("Overall health:", round(report.overall_health, 3))
```
